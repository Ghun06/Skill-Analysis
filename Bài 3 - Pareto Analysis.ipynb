```python
# Pareto Analysis
df.head()
```

```python
# Đâu là mặt hàng đáng quan tâm

### ==> Mặt hàng ? Sub Category

### Đáng quan tâm ? Total Sales cao
```

```python
# Bước 1: Tổng hợp dữ liệu

# Bước 2: Sau đó sort dữ liệu

df_sales = df.groupby(by = 'Sub Category', as_index = False) \
  .agg(total_sales = ('Sales', 'sum')) \
  .sort_values(by = 'total_sales', ascending = False) \
  .reset_index(drop = True)

df_sales.head(5)
```

```python
sum_total_sales = df_sales['total_sales'].sum()

# Tỉ trọng
df_sales['pct_sales'] = df_sales['total_sales'] * 100 / sum_total_sales
df_sales.head()
```

```python
# Lũy kế
df_sales['cum_pct_sales'] = df_sales['pct_sales'].cumsum()
df_sales.head()
```

```python
from matplotlib.patches import Patch

# --- VẼ BIỂU ĐỒ PARETO ---
fig, ax1 = plt.subplots(figsize=(12, 6))

# 1. Màu sắc theo ngưỡng 80%
colors = [
    "#FF7F0E" if pct <= 80 else "#A9A9A9"
    for pct in df_sales["cum_pct_sales"]
]

# 2. Vẽ biểu đồ cột
sub_col = (
    "Sub-Category" if "Sub-Category" in df_sales.columns else df_sales.columns[0]
)

bars = ax1.bar(
    df_sales[sub_col],
    df_sales["total_sales"],
    color=colors,
    edgecolor="black",
    linewidth=0.5,
    alpha=0.85,
)
ax1.set_xlabel("Sub-Category", fontsize=12, fontweight="bold", labelpad=10)
ax1.set_ylabel("Doanh thu (Sales)", fontsize=12, color="black", fontweight="bold")
ax1.grid(axis="y", linestyle="--", alpha=0.3)

# Căn lề phải cho nhãn trục X xoay 45 độ
plt.setp(
    ax1.get_xticklabels(), rotation=45, ha="right", rotation_mode="anchor"
)

# 3. Trục tung thứ 2 cho đường % tích lũy
ax2 = ax1.twinx()
ax2.plot(
    df_sales[sub_col],
    df_sales["cum_pct_sales"] ,
    color="#1F77B4",
    marker="o",
    linewidth=2,
    markersize=6,
    label="Tích lũy (%)",
)
ax2.set_ylabel(
    "Tỷ lệ tích lũy (%)", fontsize=12, color="#1F77B4", fontweight="bold"
)
ax2.set_ylim(0, 105)

# 4. Ngưỡng cắt 80%
ax2.axhline(
    y=80,
    color="red",
    linestyle="--",
    linewidth=1.5,
    label="Ngưỡng 80% (Pareto)",
)

# 5. Tiêu đề
plt.title(
    "BIỂU ĐỒ PARETO - PHÂN TÍCH DOANH THU THEO SUB-CATEGORY",
    fontsize=14,
    fontweight="bold",
    pad=15,
)

# --- 6. LEGEND BÊN NGOÀI GÓC TÊN PHẢI ---
legend_elements = [
    Patch(facecolor="#FF7F0E", label="Nhóm tạo 80% Doanh thu (<= 80%)"),
    Patch(facecolor="#A9A9A9", label="Nhóm còn lại (> 80%)"),
] + ax2.get_legend_handles_labels()[0]

# Đặt Legend ra ngoài bằng bbox_to_anchor
ax2.legend(
    handles=legend_elements,
    loc="upper left",
    bbox_to_anchor=(1.05, 1),
    borderaxespad=0.0,
)

plt.tight_layout()
plt.show()
```
